## testing Phase

```python
import torch
import torch.nn as nn

from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from model_ANN import ANNModel
from model_CNN import CNNModel
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using Device:", device)
```

```python
model = CNNModel().to(device)

model_path = "../models/Final Tuned CNN_cnn_opt-rmsprop_lr-0.001_drop-0.2_wd-0.0_epoch-10.pth"

model.load_state_dict(
    torch.load(model_path, map_location=device)
)

model.eval()

print("CNN Model Loaded Successfully")
```

```python
all_preds = []
all_labels = []

correct = 0
total = 0

# Disable gradient calculation
with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        # Forward pass
        outputs = model(images)

        # Get predicted class
        _, preds = torch.max(outputs, 1)

        # Store predictions
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

        # Accuracy calculation
        correct += (preds == labels).sum().item()

        total += labels.size(0)

# Final accuracy
accuracy = 100 * correct / total

print(f"\nTest Accuracy: {accuracy:.2f}%")
```

```python
print("\nClassification Report:\n")

print(classification_report(
    all_labels,
    all_preds,
    target_names=class_names
))
```

```python
cm = confusion_matrix(all_labels, all_preds)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```

### Confusion Matrix Analysis

The confusion matrix demonstrates the performance of the Final Tuned CNN model on the test dataset.

Out of 76 test images, the model correctly classified 72 images and incorrectly classified only 4 images, resulting in an overall accuracy of approximately 95%.

For the "not_ok" class, the model correctly identified 7 out of 9 samples

For the "ok" class, the model correctly classified 65 out of 67 samples
